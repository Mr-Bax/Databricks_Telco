#### Crear Catalogs y Schemas para el Proyecto
1. Catalog - Telco_dev (sin Managed Locations)
2. Schemas - bronze, silver y gold (con  Managed Location)

```sql
CREATE CATALOG if NOT EXISTS telco_dev;
```

```sql
USE CATALOG telco_dev;
```
